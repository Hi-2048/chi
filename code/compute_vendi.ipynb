```python
from vendi_score import image_utils
from PIL import Image
import os
import pandas as pd
from PIL import ImageFile
ImageFile.LOAD_TRUNCATED_IMAGES = True
```

```python
countries = ['Belgium', 'Brazil', 'Canada', 'China', 'Colombia', 'Denmark', 'Drc', 'Egypt', 'Ethiopia', 'Finland', 'France', 'Germany', 'Greece', 'India', 'Indonesia', 'Iran', 'Italy', 'Japan', 'Mexico', 'Netherlands', 'Nigeria', 'Pakistan', 'Poland', 'Russia', 'Spain', 'Sweden', 'Turkey', 'Uk', 'Usa', 'Vietnam']
classes = ['Apparel', 'Architecture', 'Artwork', 'Craft', 'Festival_activity', 'Food', 'Game_activity', 'Mythical_figure', 'Performance']
```

```python
sd15 = pd.DataFrame(columns=classes, index=countries)
sd15_dir = "sd15"
for c in classes:
    countries = os.listdir(f"{sd15_dir}/{c}")
    for ct in countries:
        current_dir = f"{sd15_dir}/{c}/{ct}"
        images = []
        for i in os.listdir(current_dir):
            if i.endswith(".png"):
                image = Image.open(f"{current_dir}/{i}").convert("RGB")
                images.append(image)
        score = image_utils.embedding_vendi_score(images, device="cuda", batch_size=len(images))
        sd15.loc[ct, c] = score
sd15.to_csv("sd15.csv")
```
